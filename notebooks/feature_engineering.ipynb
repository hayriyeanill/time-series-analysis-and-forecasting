## 🎯 Purpose

This notebook demonstrates feature engineering techniques for time series forecasting. Feature engineering transforms raw time series data into meaningful features that capture temporal patterns, improving model performance.

## 📂 Contents

1. **Time Delay Embedding**
   - **Lag Features**: Create lag features for 1 day, 1 week, and 1 month for each forecast horizon
   - **Rolling Window Features**: Generate rolling statistics (mean, std, min, max) over different windows

2. **Temporal Embedding**
   - **Calendar Features**: Extract year, month, hour, day of week, weekend indicator, and day of year
   - **Time Elapsed**: Create continuous time feature (Unix timestamp)
   - **Fourier Terms**: Generate daily and yearly Fourier terms to capture cyclical patterns

3. **Feature Selection**
   - Correlation analysis to identify most relevant features


```python
import warnings
warnings.filterwarnings("ignore")

import sys
from pathlib import Path
sys.path.append(str(Path().resolve().parent))

import pandas as pd

from data.paths import PROCESSED_DATA_DIR
from src import feature_engineering as fe 
```

```python
dataset = pd.read_csv(f"{PROCESSED_DATA_DIR}/processed_dataset.csv", index_col=0, parse_dates=True)
dataset
```

```python
dataset.info()
```

```python
forecast_days = 5
timestamps_per_day = 24 # 1 day is 24 hour
forecast_horizon = forecast_days * timestamps_per_day  # 120 time steps
```

```python
target = "temperature_2m (°C)"
data = dataset[[target]]
data = data.asfreq('h')
data
```

## Time delay embedding

### Lag values

```python
lagged_data = fe.create_lag_features(data, target, timestamps_per_day, 24)
lagged_data
```

### Rolling window values

```python
rolled_data = fe.create_rolling_features(data, target, timestamps_per_day, 24)
rolled_data
```

## Temporal embedding

### Calendar Features

```python
calendar_data = fe.create_calendar_features(data)
calendar_data
```

### Time Elapsed

```python
time_elapsed_data = fe.create_time_elapsed_feature(data)
time_elapsed_data
```

### Fourier terms

```python
daily_fourier_data = fe.create_daily_fourier_terms(data)
daily_fourier_data
```

```python
yearly_fourier_data = fe.create_yearly_fourier_terms(data)
yearly_fourier_data
```

## Feature Selection - Correlation Analysis

```python
lagged_corr = lagged_data.corr()[target].sort_values(ascending=False)
lagged_corr
```

```python
rolling_corr = rolled_data.corr()[target].sort_values(ascending=False)
rolling_corr
```

```python
calendar_data_corr = calendar_data.corr()[target].sort_values(ascending=False)
calendar_data_corr
```

```python
time_elapsed_data.corr()[target].sort_values(ascending=False)
```

```python
fourier_data = pd.merge(daily_fourier_data, yearly_fourier_data.iloc[:,1:], left_index=True, right_index=True)
fourier_data_corr = fourier_data.corr()[target].sort_values(ascending=False)
fourier_data_corr
```
